```python
task_id = list(train_tasks.keys())[1]

print("task_id:", task_id)
print(train_tasks[task_id])


def solve(grid):
    n = len(grid)

   
    result = [[0] * (n * n) for _ in range(n * n)]

   
    for i in range(n):
        for j in range(n):

            if grid[i][j] == 0:
                continue

            
            for r in range(n):
                for c in range(n):
                    result[i*n + r][j*n + c] = grid[r][c]

    return result



for example in train_tasks[task_id]['train']:

    inp = example['input']
    out = example['output']

    pred = solve(inp)

    print(pred == out)

test_input = train_tasks[task_id]['test'][0]['input']

answer = solve(test_input)

print(answer)
```

```python
task_id = list(train_tasks.keys())[2]

print("task_id:", task_id)

task = train_tasks[task_id]


def solve(grid):
    h = len(grid)
    w = len(grid[0])

    shape_to_color = {
        ((1,1,1),
         (1,0,1),
         (0,1,0)): 7,

        ((1,0,1),
         (0,1,0),
         (1,1,1)): 3,

        ((0,1,0),
         (1,1,1),
         (0,1,0)): 2
    }

    ones = []

    for r in range(h):
        for c in range(w):
            if grid[r][c] == 1:
                ones.append((r, c))

    min_r = min(r for r, c in ones)
    max_r = max(r for r, c in ones)
    min_c = min(c for r, c in ones)
    max_c = max(c for r, c in ones)

    shape = []

    for r in range(min_r, max_r + 1):
        row = []

        for c in range(min_c, max_c + 1):
            if grid[r][c] == 1:
                row.append(1)
            else:
                row.append(0)

        shape.append(tuple(row))

    shape = tuple(shape)

    new_color = shape_to_color[shape]

    result = [[0] * w for _ in range(h)]

    for r in range(h):
        for c in range(w):
            if grid[r][c] == 8:
                result[r][c] = new_color

    return result


test_input = task["test"][0]["input"]

answer = solve(test_input)

for row in answer:
    print(row)
```
