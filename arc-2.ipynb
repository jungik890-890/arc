```python
task_id = list(train_tasks.keys())[1]

print("task_id:", task_id)
print(train_tasks[task_id])


def solve(grid):
    n = len(grid)

   
    result = [[0] * (n * n) for _ in range(n * n)]

   
    for i in range(n):
        for j in range(n):

            if grid[i][j] == 0:
                continue

            
            for r in range(n):
                for c in range(n):
                    result[i*n + r][j*n + c] = grid[r][c]

    return result



for example in train_tasks[task_id]['train']:

    inp = example['input']
    out = example['output']

    pred = solve(inp)

    print(pred == out)

test_input = train_tasks[task_id]['test'][0]['input']

answer = solve(test_input)

print(answer)
```
